# Import

```python
from google.colab import files
from PIL import Image
```

# Image upload

```python
u = files.upload()
```

# Read Image

```python
n = list(u.keys())[0]
```

```python
im = Image.open(n).convert("RGB")
```

```python
w, h = im.size
```

```python
display(im)
```

# Get pixels

```python
p = []
for y in range(h):
    r = []
    for x in range(w):
        r.append(im.getpixel((x, y)))
    p.append(r)
```

# Increase Brightness

```python
b = 50
bp = []
for y in range(h):
    r = []
    for x in range(w):
        R, G, B = p[y][x]
        R = R + b
        G = G + b
        B = B + b
        if R > 255:
            R = 255
        if G > 255:
            G = 255
        if B > 255:
            B = 255
        r.append((R, G, B))
    bp.append(r)
```

# Display bright image

```python
bi = Image.new("RGB", (w, h))
for y in range(h):
    for x in range(w):
        bi.putpixel((x, y), bp[y][x])
display(bi)
```

# Decrease Image

```python
b = 50
dp = []
for y in range(h):
    r = []
    for x in range(w):
        R, G, B = p[y][x]
        R = R - b
        G = G - b
        B = B - b
        if R < 0:
            R = 0
        if G < 0:
            G = 0
        if B < 0:
            B = 0
        r.append((R, G, B))
    dp.append(r)
```

# Display Dark image

```python
di = Image.new("RGB", (w, h))
for y in range(h):
    for x in range(w):
        di.putpixel((x, y), dp[y][x])
display(di)
```
